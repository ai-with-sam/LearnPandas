```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry", "Ivy", "Jack"],
    "Age": [25, None, 29, 31, 27, None, 30, 26, 28, 32],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", None, "HR", "Finance"],
    "Salary": [35000, 50000, None, 60000, 42000, 48000, 55000, 39000, 43000, 65000],
    "Experience": [2, 5, 4, 7, 3, 4, 6, 2, None, 8],
    "City": ["Chennai", "Bangalore", "Hyderabad", "Chennai", "Mumbai", "Delhi", "Chennai", "Bangalore", "Hyderabad", "Delhi"],
    "Phone": [
        "9876543210", "9876543211", "9876543212", None,
        "9876543214", "9876543215", "9876543216",
        "9876543217", "9876543218", "9876543219"
    ]
})

df
```

```python
# Task 1

# Display the first five rows of the DataFrame.

df.head()
```

```python
# Task 2

# Remove the Phone column.

df.drop(columns='Phone', inplace=True)
df
```

```python
# Task 3

# Find the total number of missing values in each column.

df.isna().sum()
```

```python
# Task 4

# Replace all missing values using appropriate values based on the column.
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Experience'] = df['Experience'].fillna(df['Experience'].mean())

df = df.round({
    'Salary': 2,
    'Experience': 2,
    'Age': 2
})

df
```

Task 5

Rename the column Employee_ID to Emp_ID.

```python
df = df.rename(columns={"Employee_ID": "Emp_ID"})
df
```
