```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Alice", "Bob", "Charlie", "David", "Eva",
        "Frank", "Grace", "Henry", "Ivy", "Jack"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "HR", "IT", "Finance"
    ],
    "Age": [25, 30, 28, 35, 27, 32, 29, 31, 26, 38],
    "Salary": [
        55000, 42000, 68000, 72000, 45000,
        60000, 65000, 48000, 58000, 75000
    ],
    "Experience": [2, 5, 3, 8, 4, 7, 5, 6, 2, 10]
})

df
```

Task 1

Create a NumPy array containing the salaries from the DataFrame.

```python
sal_array = np.array(df['Salary'])
sal_array
```

Task 2

Using NumPy, calculate the average salary.

```python
avg_sal = np.mean(sal_array)
avg_sal
```

Task 3

Using NumPy, find the highest salary and lowest salary.

```python
max_sal = np.max(sal_array)
min_sal = np.min(sal_array)

print(f"Max salary: {max_sal}")
print(f"Min_salary: {min_sal}")
```

Task 4

Create a new column named Salary_In_Lakhs by converting the salary from rupees to lakhs.

For example:

₹55000 → 0.55

```python
df["Salary_In_Lakhs"] = df["Salary"] / 100000

print(df)
```

Task 5

Create a new column named Experience_Level based on the employee's experience:

Experience less than 3 years → "Beginner"
Experience from 3 to 6 years → "Intermediate"
Experience greater than 6 years → "Experienced"

```python
df["Experience_Level"] = np.where(
    df["Experience"] < 3,
    "Beginner",
    np.where(
        df["Experience"] <= 6,
        "Intermediate",
        "Experienced"
    )
)

df
```
