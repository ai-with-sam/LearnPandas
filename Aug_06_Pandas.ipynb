```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    "Name": [" Alice ", "bob", "CHARLIE", "David", "eva ", " Frank", "Grace", "henry", "IVY", "Jack"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR", "IT", "Finance"],
    "Gender": ["M", "F", "M", "M", "F", "M", "F", "M", "F", "M"],
    "Salary": [50000, 42000, 61000, 58000, 45000, 52000, 60000, 43000, 62000, 59000],
    "City": ["Chennai", "Bangalore", "Chennai", "Delhi", "Mumbai", "Chennai", "Delhi", "Bangalore", "Mumbai", "Delhi"]
})

df
```

Task 1

Display the number of employees in each department.

```python
df['Department'].value_counts()
```

Task 2

Display all unique city names.

```python
df['City'].unique()
```

Task 3

Find the total number of unique cities.

```python
df['City'].nunique()
```

Task 4

Replace all values in the Gender column:

M → Male
F → Female

```python
rep = {
    'M':'Male',
    'F':'Female'
}
df['Gender'] = df['Gender'].replace(rep)
df
```

Task 5

Clean the Name column by:

Removing leading/trailing spaces.
Converting all names to uppercase.

```python
df['Name'] = df['Name'].str.strip().str.upper()
df
```
