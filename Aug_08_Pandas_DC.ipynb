```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 107, 109, 110],
    "Name": [
        " Alice ", "BOB", "charlie", " David", "EVA ",
        "Frank", "Grace", "Grace", " Henry ", "Ivy"
    ],
    "Age": ["25", "30", "28", "35", "22", "29", "31", "31", "27", "33"],
    "Gender": [
        "Male", "Female", "M", "Female", "F",
        "Male", "M", "M", "Female", "F"
    ],
    "City": [
        "Chennai", "Bangalore", "chennai", "Delhi", "Mumbai",
        "BANGALORE", "Delhi", "Delhi", "mumbai", "Chennai"
    ],
    "Purchase_Amount": [
        2500, 1800, 3200, 4500, 2100,
        3800, 5200, 5200, 2900, 4100
    ]
})

df
```

Task 1

Check whether the DataFrame contains any duplicate rows.

```python
print(df.duplicated().sum())
```

Task 2

Remove the duplicate rows from the DataFrame.

```python
df = df.drop_duplicates()
df
```

Task 3

Standardize the Gender column so that it contains only:

Male
Female

```python
rep = {
    'M': 'Male',
    'F': 'Female'
}

df['Gender'] = df["Gender"].replace(rep)
df
```

Task 4

Standardize the City column so that all city names follow the same capitalization format.

For example, the final values should look like:

Chennai
Bangalore
Delhi
Mumbai

```python
df['City'] = df['City'].str.capitalize()
df
```

Task 5

Clean the Name column so that every name:

Has no unnecessary spaces.
Starts with an uppercase letter.
Has the remaining letters in lowercase.

For example:

" Alice " → "Alice"
"BOB" → "Bob"
"charlie" → "Charlie"

```python
df['Name'] = df['Name'].str.title()
df
```
