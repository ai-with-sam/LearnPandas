```python
import pandas as pd

df = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 107, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Grace", "Ivy", "Jack"],
    "Age": [21, 22, None, 24, 20, 23, 22, 22, None, 25],
    "Course": ["Python", "Java", "Python", "SQL", None, "Python", "Java", "Java", "SQL", "Python"],
    "Marks": [85, 78, 92, None, 88, 75, 81, 81, 90, 95],
    "City": ["Chennai", "Bangalore", "Hyderabad", "Chennai", "Mumbai", "Delhi", "Chennai", "Chennai", "Hyderabad", "Delhi"]
})

df
```

```python
#1. Display only the students whose Marks are greater than or equal to 85.
df[df["Marks"] >= 85]["Name"]

# if u want in list
# df[df["Marks"] >= 85]["Name"].toList()
```

```python
#2. Replace all missing values in the DataFrame with appropriate values.

df['Age'] = df["Age"].fillna(round(df["Age"].mean(),2))
df['Course'] = df['Course'].fillna(df['Course'].mode()[0])
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

df
```

```python
#3. Remove all duplicate rows from the DataFrame
df = df.drop_duplicates()
df
```

```python
#4. Sort the DataFrame by Marks in descending order.
df = df.sort_values(by="Marks", ascending = False)
df
```

```python
#5. Display only the Name, Course, and Marks columns.
df[['Name','Course', 'Marks']]
```
