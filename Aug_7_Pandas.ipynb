```python
import pandas as pd

df = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry", "Ivy", "Jack"],
    "Category": ["Electronics", "Clothing", "Electronics", "Furniture", "Clothing", "Furniture", "Electronics", "Clothing", "Furniture", "Electronics"],
    "Price": ["25000", "1800", "32000", "15000", "2200", "18000", "28000", "2500", "17000", "35000"],
    "Quantity": [1, 3, 1, 2, 4, 1, 2, 5, 2, 1],
    "City": ["Chennai", "Delhi", "Mumbai", "Chennai", "Hyderabad", "Delhi", "Mumbai", "Chennai", "Hyderabad", "Delhi"],
    "Discount": [10, 5, 15, 8, 5, 10, 12, 5, 7, 20]
})

df
```

Task 1

Convert the Price column from string to integer.

```python
df["Price"] = df["Price"].astype(int)
df.info()
```

Task 2

Create a new column named Total_Amount using:

Price × Quantity

```python
df['Total_Amount'] = df['Price'] * df['Quantity']
df
```

Task 3

Display only the orders where:

Category is Electronics
Total_Amount is greater than 30000

```python
df[(df["Category"] == "Electronics") & (df["Total_Amount"] > 30000)]
```

Task 4

Sort the DataFrame by Total_Amount in descending order.

```python
df = df.sort_values(by=['Total_Amount'], ascending=False)
df
```

Task 5

Remove the Discount column.

```python
df = df.drop(columns='Total_Amount', axis=1)
df
```
