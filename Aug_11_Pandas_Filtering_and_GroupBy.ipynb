```python
import pandas as pd

df = pd.DataFrame({
    "Product_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Headphones", "Monitor",
        "Keyboard", "Laptop", "Headphones", "Mouse"
    ],
    "Category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Accessories", "Electronics", "Accessories", "Accessories"
    ],
    "Price": [
        65000, 1200, 2500, 18000,
        72000, 1500, 4500, 22000,
        2800, 68000, 5000, 1300
    ],
    "Quantity_Sold": [
        5, 20, 15, 8,
        3, 25, 12, 6,
        18, 4, 10, 30
    ],
    "City": [
        "Chennai", "Delhi", "Mumbai", "Chennai",
        "Bangalore", "Delhi", "Mumbai", "Bangalore",
        "Chennai", "Delhi", "Mumbai", "Chennai"
    ]
})

df
```

Task 1

Create a new column named Revenue.

Its value should represent the revenue generated by each product.

```python
df['Revenue'] = df['Price']*df['Quantity_Sold']
df
```

Task 2

Display all products where Revenue is greater than 50,000.

```python
df[df['Revenue']>50000]
```

Task 3

Find the total Revenue generated by each Category.

```python
df.groupby('Category')['Revenue'].sum()
```

Task 4

Find the average Revenue generated in each City.

```python
df.groupby('City')['Revenue'].mean()
```

Task 5

Sort the DataFrame by Revenue, from highest to lowest.

```python
df = df.sort_values(by='Revenue', ascending=False)
df
```
