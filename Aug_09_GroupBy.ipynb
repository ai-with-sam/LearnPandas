```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Name": [
        "Alice", "Bob", "Charlie", "David", "Eva", "Frank",
        "Grace", "Henry", "Ivy", "Jack", "Kevin", "Laura"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR", "IT",
        "Finance", "HR", "IT", "Finance", "HR", "IT"
    ],
    "Age": [25, 30, 28, 35, 27, 32, 29, 31, 26, 38, 24, 33],
    "Salary": [
        55000, 42000, 68000, 72000, 45000, 60000,
        65000, 48000, 58000, 75000, 40000, 70000
    ],
    "Experience": [2, 5, 3, 8, 4, 7, 5, 6, 2, 10, 1, 8],
    "City": [
        "Chennai", "Bangalore", "Chennai", "Delhi",
        "Mumbai", "Chennai", "Delhi", "Bangalore",
        "Chennai", "Delhi", "Mumbai", "Chennai"
    ]
})

df
```

Task 1

Find the average salary for each department.

```python
df.groupby('Department')['Salary'].mean()
```

Task 2

Find the maximum salary for each department.

```python
df.groupby('Department')['Salary'].max()
```

Task 3

Find the number of employees in each department.

```python
df.groupby('Department')['Employee_ID'].count()
```

Task 4

Find the average experience for each city.

```python
df.groupby('City')['Experience'].mean()
```

Task 5

Find the total salary paid by each department.

```python
df.groupby('Department')['Salary'].sum()
```

⭐ Bonus Challenge

Create a summary table containing each department with:

Number of employees
Average salary
Maximum salary
Minimum salary

The final result should have one row per department.

```python
summary = df.groupby("Department").agg(
    Employee_Count=("Employee_ID", "count"),
    Average_Salary=("Salary", "mean"),
    Maximum_Salary=("Salary", "max"),
    Minimum_Salary=("Salary", "min")
)

print(summary)
```
